```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

x = np.linspace(-5, 5, 200)
plt.plot(x, x**2)
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-3, 4, 300)
y = (x + 1) * (x - 2)* (x - 4)

plt.plot(x, y)
plt.axhline(0, color='black', linewidth=0.8)
plt.scatter([-1, 2, 4], [0, 0, 0], color='red', zorder=5, s=60)
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-3, 4, 300)
y = (x + 2) * (x - 1) * (x - 3)

plt.plot(x, y)
plt.axhline(0, color='black', linewidth=0.8)

for r in [-2, 1, 3]:
    plt.axvline(r, color='red', linestyle='--', linewidth=1)

plt.grid(True)
plt.show()
```

```python
x = np.linspace(-2, 6, 300)
y = x**2 - 4*x + 3          # (x-1)(x-3)

plt.plot(x, y)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(2, color='green', linestyle='--', label='axis: x=2')

for r in [1, 3]:
    plt.axvline(r, color='red', linestyle=':', linewidth=1)

plt.scatter([2], [-1], color='green', zorder=5)
plt.legend()
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-2, 6, 300)
y = x**2 - 4*x + 3

plt.plot(x, y)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvspan(1, 3, color='orange', alpha=0.2)   # 두 근 사이

plt.grid(True)
plt.show()
```

```python
a = np.linspace(-3, 3, 200)
b = 2

left = (a+b)**2
right = a**2+2*a*b+b**2

plt.plot(a, left, linewidth=4, alpha=0.4, label='(a+b)^2')
plt.plot(a, right, linewidth=1.5, label='a^2+2ab+b^2')
plt.legend(); plt.grid(True); plt.show()
```

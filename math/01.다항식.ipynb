```python
x = np.linspace(-2, 6, 300)
y = x**2 - 4*x + 3          # (x-1)(x-3)

plt.plot(x, y)
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(2, color='green', linestyle='--', label='axis: x=2')

for r in [1, 3]:
    plt.axvline(r, color='red', linestyle=':', linewidth=1)

plt.scatter([2], [-1], color='green', zorder=5)
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def P(x):
    return x**3 - 6*x**2 + 11*x -6

# for a in [-1, 0, 1, 2, 3, 4]:
#     print(f"P({a})={P(a)}")

x = np.linspace(0, 4, 300)
plt.plot(x, P(x))
plt.axhline(0, color='black', linewidth=0.8)

for a in [1, 2, 3]:
    plt.axvline(a, color='red', linestyle='--', linewidth=1)

plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 4, 300)

original = P(x)
factored = (x-1) * (x-2) * (x-3)

plt.plot(x, original, linewidth=5, alpha=0.3, label='original')
plt.plot(x, factored, linewidth=1.5, label='factored')
plt.legend(); plt.grid(True); plt.show()
```

```python
from numpy.polynomial import polynomial as Poly

# 계수를 상수항부터 사용, 프로그래밍에서는 상수항부터 
p= [-6, 11, -6, 1]

for a in [1, 2, 4, 5]:
    몫, 나머지 = Poly.polydiv(p, [-a, 1]) # polydiv는 결과를 두개, 한번에 두 변수로 받는걸 언패킹

    print(f"a={a}, 나머지={나머지}, P({a})={np.polyval(p[::-1], a)}")

#polydiv는 상수항부터, polyval는 최고차부터
```

```python
import numpy as np

def f(x):
    return x**3 - 2*x**2 - 5*x + 6

def 약수(n):
    n = abs(n)
    return [d for d in range(1, n+1) if n % d == 0]

상수항, 최고차 = 6, 1

후보 = sorted({s * p / q for p in 약수(상수항) for q in 약수(최고차) for s in (1, -1)})
print("후보:", 후보)

for c in 후보:
    if abs(f(c)) < 1e-10:
        print(f"근: {c}")
        
```

```python
import matplotlib.pyplot as plt

x = np.linspace(-3, 4, 300)
plt.plot(x, f(x))
plt.axhline(0, color='black', linewidth=0.8)

for r in [-2, 1, 3]:
    plt.axvline(r, color='red', linestyle='--', linewidth =1)

plt.grid(True)
plt.show()
```
